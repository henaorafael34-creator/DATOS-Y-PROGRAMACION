#### 3. Exploración DFS (Búsqueda en Profundidad / Exploración de cuevas - Punto 3)

***Explicación técnica:** DFS explora un camino tan lejos como sea posible a lo largo de cada rama antes de retroceder (*backtracking*). Utiliza una estructura **LIFO (Pila / Stack)** o la pila de llamadas mediante **recursividad**, asegurando que cada nodo se procese una sola vez mediante el control de `visitados`.

**En Minecraft:** Es la típica táctica de picar hacia abajo (*minería en profundidad*) o explorar una cueva enorme hacia el fondo: sigues un túnel a fondo hasta chocar con roca madre o lava (nodo sin más vecinos sin visitar) y solo ahí te devuelves (*backtracking*) para tomar el otro camino que dejaste atrás.

**En Minecraft:** Funciona exactamente como usar una **brújula de magnetita** o los rieles más rápidos del servidor para ir desde tu base inicial hasta la base del *Ender Dragon*. Busca el trayecto que te tome la menor cantidad de portales del Nether o vagonetas posibles.

### Quinta Parte: Análisis de Complejidad Algorítmica

En la teoría de grafos, cuando representamos la red con una **lista de adyacencia**, el tamaño de la entrada depende de dos variables principales:

* $n = \vert{}V\vert{}$: Número de vértices (usuarios).


* $m = \vert{}E\vert{}$: Número de aristas (conexiones/amistades).


### 1. Construcción y Muestra de la Lista de Adyacencia (Punto 1)

* **Función de Tiempo $T(n, m)$:**

$$T(n, m) = c_1 \cdot n + c_2 \cdot m$$

* **Justificación:** Para inicializar y mostrar la lista, el algoritmo debe iterar sobre los $n$ usuarios y, para cada uno, recorrer la lista de sus $m$ vecinos totales (en un grafo no dirigido, la suma de los grados de todos los nodos es $2m$).
* **Complejidad Asintótica:** $\mathcal{O}(n + m)$ (Lineal respecto al tamaño del grafo).
* **Casos:**
* **Mejor Caso $\Omega(n)$:** Grafo completamente disperso sin conexiones ($m = 0$). Solo se iteran los $n$ nodos.
* **Peor Caso $\mathcal{O}(n^2)$:** Grafo completo (todos son amigos de todos, $m = \frac{n(n-1)}{2}$).


### 2. Recorrido BFS y Cálculo de Grados de Separación (Punto 2 y 5)

* **Función de Tiempo $T(n, m)$:**

$$T(n, m) = c_1 + c_2 \cdot n + c_3 \cdot m$$

* **Justificación:**
1. Cada vértice entra y sale de la cola exactamente una sola vez gracias al control de visitados $\rightarrow \mathcal{O}(n)$.

2. Para cada vértice desencolado, se examinan todas sus aristas salientes $\rightarrow \sum \text{grado}(v) = 2m \rightarrow \mathcal{O}(m)$.

* **Complejidad Asintótica:** $\mathcal{O}(n + m)$.
* **Casos:**
* **Mejor Caso $\Omega(1)$ / $\Omega(n)$:** Si el nodo origen no tiene conexiones, solo se procesa él mismo en $\mathcal{O}(1)$. Si se busca un destino y este es el vecino inmediato del origen, BFS finaliza tempranamente en $\mathcal{O}(1)$.
* **Peor Caso $\mathcal{O}(n + m)$:** El nodo destino está en el extremo más lejano de la red o el grafo está completamente conectado y se deben recorrer todos los nodos y aristas.

### 3. Recorrido DFS (Punto 3)

* **Función de Tiempo $T(n, m)$:**

$$T(n, m) = c_1 \cdot n + c_2 \cdot m$$

* **Justificación:** Al igual que BFS, el uso del conjunto de `visitados` garantiza que cada vértice sea procesado únicamente una vez (ya sea mediante la pila o recursividad) y que cada arista sea explorada una sola vez por cada extremo.
* **Complejidad Asintótica:** $\mathcal{O}(n + m)$.
* **Casos:**
* **Mejor Caso $\Omega(1)$:** El nodo inicial no tiene aristas, por lo que la pila/recursividad termina inmediatamente.
* **Peor Caso $\mathcal{O}(n + m)$:** Toda la componente conexa debe ser explorada en profundidad hasta el último nivel.


### 4. Búsqueda y Reconstrucción de la Ruta Más Corta (Punto 5)

* **Función de Tiempo $T(n, m)$:**

$$T(n, m) = T_{\text{BFS}}(n, m) + T_{\text{Reconstrucción}}(k) = (c_1 \cdot n + c_2 \cdot m) + c_3 \cdot k$$

*(donde $k \le n$ es la longitud del camino encontrado en saltos)*.
* **Justificación:** Se ejecuta una exploración BFS para encontrar el objetivo y registrar los nodos padres. Una vez hallado el destino, el bucle de reconstrucción de la ruta toma $k$ pasos (recorriendo hacia atrás desde el destino al origen). Como $k \le n$, el término dominante sigue siendo BFS.
* **Complejidad Asintótica:** $\mathcal{O}(n + m)$.
* **Casos:**
* **Mejor Caso $\Omega(1)$:** El origen y el destino son la misma persona o son amigos directos (1 solo salto).
* **Peor Caso $\mathcal{O}(n + m)$:** El camino requiere atravesar casi todos los nodos del grafo o evaluar todas las ramas antes de encontrar el destino.


### 5. Inserción de un Nuevo Usuario y Conexiones (Punto 6)

* **Función de Tiempo $T(k)$:**

$$T(k) = c_1 + c_2 \cdot k$$

*(donde $k$ es el número de conexiones que se le asignan al nuevo usuario; en el taller $k \ge 2$)*.

* **Justificación:** Crear la clave del nuevo usuario en la lista de adyacencia toma tiempo constante $\mathcal{O}(1)$. Agregar cada una de sus $k$ amistades toma $\mathcal{O}(1)$ para insertarlo en la lista del nuevo usuario y $\mathcal{O}(1)$ para insertarlo en la lista de cada amigo existente.
* **Complejidad Asintótica:** $\mathcal{O}(k)$, que en la práctica con $k \ge 2$ pequeño equivale a **$\mathcal{O}(1)$ (Tiempo constante)**.
* **Casos:**
* **Mejor Caso $\Omega(1)$:** Se conecta con el mínimo de 2 personas.
* **Peor Caso $\mathcal{O}(n)$:** El nuevo usuario se conecta directamente con los $n$ usuarios preexistentes de la red.

```python
class EstadisticasRed:
    """Clase auxiliar para calcular métricas y metadatos avanzados de la red."""
    def __init__(self, grafo):
        self.grafo = grafo
 
    def total_usuarios(self):
        return len(self.grafo)
 
    def total_amistades(self):
        # Al ser grafo no dirigido, la suma de grados se divide entre 2
        return sum(len(amigos) for amigos in self.grafo.values()) // 2
 
    def usuario_mas_influyente(self):
        if not self.grafo:
            return None, 0
        usuario_top = max(self.grafo.items(), key=lambda x: len(x[1]))
        return usuario_top[0], len(usuario_top[1])
 
 
class RedSocialAvanzada:
    """Sistema integral interactivo de red social basado en grafos no dirigidos."""
    
    def __init__(self):
        self.grafo = {}
        self.stats = EstadisticasRed(self.grafo)
        self._cargar_red_inicial()
 
    def _cargar_red_inicial(self):
        """Carga la red base obligatoria del taller."""
        usuarios_base = ["Ana", "Carlos", "Laura", "Mateo", "Sofía", "Daniel", "Valentina"]
        for u in usuarios_base:
            self.grafo[u] = []
            
        conexiones_base = [
            ("Ana", "Carlos"),
            ("Ana", "Laura"),
            ("Carlos", "Mateo"),
            ("Carlos", "Sofía"),
            ("Laura", "Daniel"),
            ("Sofía", "Valentina"),
            ("Daniel", "Valentina")
        ]
        
        for u1, u2 in conexiones_base:
            self._conectar_bidireccional(u1, u2)
 
    def _conectar_bidireccional(self, u1, u2):
        """Garantiza la simetría de las aristas (grafo no dirigido) sin duplicados."""
        if u1 in self.grafo and u2 in self.grafo:
            if u2 not in self.grafo[u1]:
                self.grafo[u1].append(u2)
            if u1 not in self.grafo[u2]:
                self.grafo[u2].append(u1)
 
    # -------------------------------------------------------------------------
    # PUNTO 1: REPRESENTACIÓN DE LA RED SOCIAL
    # -------------------------------------------------------------------------
    def mostrar_red(self):
        print("\n" + "=" * 65)
        print(" [PUNTO 1] REPRESENTACIÓN DE LA RED SOCIAL (LISTA DE ADYACENCIA)")
        print("=" * 65)
        for usuario, amigos in self.grafo.items():
            amigos_str = ", ".join(amigos) if amigos else "Sin conexiones"
            print(f" 👤 Nodo [{usuario}] ──► Conectado con: [{amigos_str}]")
        print("-" * 65)
        print(f" 📊 Estadísticas: Total de usuarios (n) = {self.stats.total_usuarios()} | Total de amistades (m) = {self.stats.total_amistades()}")
        top_user, top_amigos = self.stats.usuario_mas_influyente()
        print(f" 🔥 Usuario más conectado: {top_user} ({top_amigos} conexiones)")
        print("=" * 65 + "\n")
 
    # -------------------------------------------------------------------------
    # PUNTO 2: EXPLORACIÓN BFS (BÚSQUEDA EN ANCHURA)
    # -------------------------------------------------------------------------
    def bfs(self, origen):
        if origen not in self.grafo:
            print(f"\n[ERROR DE SISTEMA] El usuario '{origen}' no existe en la red.\n")
            return None
            
        visitados = {origen}
        grados_separacion = {origen: 0}
        cola = [origen]
        orden_visita = []
        
        while cola:
            actual = cola.pop(0)
            orden_visita.append(actual)
            
            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    grados_separacion[vecino] = grados_separacion[actual] + 1
                    cola.append(vecino)
                    
        print(f"\n--- RECORRIDO BFS (ANCHO) DESDE: '{origen}' ---")
        print(f"Orden de visita: {' -> '.join(orden_visita)}")
        
        print("\nGrados de separación por niveles:")
        niveles = {}
        for nodo, grado in grados_separacion.items():
            niveles.setdefault(grado, []).append(nodo)
        for grado in sorted(niveles.keys()):
            print(f" • Grado {grado}: {', '.join(niveles[grado])}")
        print("-" * 50 + "\n")
        return grados_separacion
 
    # -------------------------------------------------------------------------
    # PUNTO 3: EXPLORACIÓN DFS (BÚSQUEDA EN PROFUNDIDAD)
    # -------------------------------------------------------------------------
    def dfs(self, origen):
        if origen not in self.grafo:
            print(f"\n[ERROR DE SISTEMA] El usuario '{origen}' no existe en la red.\n")
            return
            
        visitados = set()
        orden_visita = []
        
        def _recorrido_recursivo(nodo):
            visitados.add(nodo)
            orden_visita.append(nodo)
            for vecino in self.grafo[nodo]:
                if vecino not in visitados:
                    _recorrido_recursivo(vecino)
                    
        _recorrido_recursivo(origen)
        print(f"\n--- RECORRIDO DFS (PROFUNDIDAD) DESDE: '{origen}' ---")
        print(f"Un posible recorrido: {' -> '.join(orden_visita)}")
        print("-" * 50 + "\n")
 
    # -------------------------------------------------------------------------
    # PUNTO 4: MENÚ DINÁMICO DE SELECCIÓN DE USUARIO INICIAL
    # -------------------------------------------------------------------------
    def menu_seleccion_usuario(self):
        print("\n" + "=" * 50)
        print(" [PUNTO 4] SELECCIÓN DINÁMICA DE USUARIO INICIAL")
        print("=" * 50)
        print(f"Usuarios actuales disponibles: {list(self.grafo.keys())}")
        usuario = input("Escriba el nombre del usuario a consultar: ").strip()
        
        if usuario not in self.grafo:
            print(f"\n[❌ ERROR] El usuario '{usuario}' no se encuentra registrado en el sistema.\n")
            return
            
        print(f"\n[✔ ÉXITO] Analizando la red desde el nodo de '{usuario}'...")
        self.bfs(usuario)
        self.dfs(usuario)
 
    # -------------------------------------------------------------------------
    # PUNTO 5: BÚSQUEDA DE RUTA MÁS CORTA ENTRE DOS PERSONAS
    # -------------------------------------------------------------------------
    def menu_ruta_minima(self):
        print("\n" + "=" * 50)
        print(" [PUNTO 5] BÚSQUEDA DE RUTA MÍNIMA (SALTOS DE AMISTAD)")
        print("=" * 50)
        print(f"Usuarios actuales: {list(self.grafo.keys())}")
        origen = input("Ingrese el usuario de inicio: ").strip()
        destino = input("Ingrese el usuario de destino: ").strip()
        
        if origen not in self.grafo or destino not in self.grafo:
            print("\n[❌ ERROR] Uno o ambos usuarios ingresados no existen en la red.\n")
            return
            
        cola = [origen]
        visitados = {origen}
        padres = {origen: None}
        encontrado = False
        
        while cola:
            actual = cola.pop(0)
            if actual == destino:
                encontrado = True
                break
            for vecino in self.grafo[actual]:
                if vecino not in visitados:
                    visitados.add(vecino)
                    padres[vecino] = actual
                    cola.append(vecino)
                    
        print(f"\n--- RESULTADO DE BÚSQUEDA DE RUTA ---")
        if not encontrado:
            print(f"No existe ninguna ruta de conexión entre '{origen}' y '{destino}'.\n")
        else:
            camino = []
            curr = destino
            while curr is not None:
                camino.append(curr)
                curr = padres[curr]
            camino.reverse()
            
            print(f"Ruta óptima encontrada: {' -> '.join(camino)}")
            print(f"Grados de separación totales: {len(camino) - 1}\n")
 
    # -------------------------------------------------------------------------
    # PUNTO 6: AGREGAR NUEVO USUARIO Y SUS CONEXIONES
    # -------------------------------------------------------------------------
    def menu_agregar_usuario(self):
        print("\n" + "=" * 50)
        print(" [PUNTO 6] AGREGAR NUEVO USUARIO A LA RED")
        print("=" * 50)
        nuevo_usuario = input("Ingrese el nombre del nuevo usuario (ej. Juliana): ").strip()
        
        if not nuevo_usuario:
            print("\n[❌ ERROR] El nombre no puede estar vacío.\n")
            return
            
        if nuevo_usuario in self.grafo:
            print(f"\n[⚠️ ADVERTENCIA] El usuario '{nuevo_usuario}' ya habita en la red.\n")
            return
            
        print(f"Usuarios existentes para conectar: {list(self.grafo.keys())}")
        c1 = input("Ingrese la primera conexión existente: ").strip()
        c2 = input("Ingrese la segunda conexión existente: ").strip()
        
        if c1 not in self.grafo or c2 not in self.grafo:
            print("\n[❌ ERROR] Ambas conexiones deben ser usuarios válidos que ya existan en la red.\n")
            return
            
        # Registrar nodo e interconectar
        self.grafo[nuevo_usuario] = []
        self._conectar_bidireccional(nuevo_usuario, c1)
        self._conectar_bidireccional(nuevo_usuario, c2)
        
        print(f"\n[🎉 ¡ÉXITO!] Usuario '{nuevo_usuario}' registrado y conectado con '{c1}' y '{c2}'.\n")
 
    # -------------------------------------------------------------------------
    # BUSCADOR DETALLADO DE UN USUARIO ESPECÍFICO
    # -------------------------------------------------------------------------
    def buscar_detalles_usuario(self):
        print("\n" + "=" * 50)
        print(" 🔍 BUSCADOR INTERACTIVO DE USUARIO ESPECÍFICO")
        print("=" * 50)
        usuario = input("Escriba el nombre del usuario a buscar: ").strip()
        
        if usuario not in self.grafo:
            print(f"\n[❌] El usuario '{usuario}' no está registrado en la base de datos de la red.\n")
            return
            
        amigos = self.grafo[usuario]
        print(f"\n📄 Ficha del usuario: {usuario}")
        print(f" • Total de amigos directos (Grado): {len(amigos)}")
        print(f" • Lista de amistades directas: {', '.join(amigos) if amigos else 'Ninguna'}")
        print("-" * 50 + "\n")
 
 
# ==============================================================================
# BUCLE PRINCIPAL DEL SISTEMA INTERACTIVO (CONSOLA / JUPYTER)
# ==============================================================================
def iniciar_consola_sistema():
    app = RedSocialAvanzada()
    
    while True:
        print("╔══════════════════════════════════════════════════════╗")
        print("║       SISTEMA DE GESTIÓN Y ANÁLISIS DE RED SOCIAL    ║")
        print("╚══════════════════════════════════════════════════════╝")
        print(" 1. Mostrar red social completa [Punto 1]")
        print(" 2. Ejecutar BFS desde 'Ana' [Punto 2]")
        print(" 3. Ejecutar DFS desde 'Ana' [Punto 3]")
        print(" 4. Seleccionar usuario inicial de forma dinámica [Punto 4]")
        print(" 5. Buscar ruta más corta entre dos usuarios [Punto 5]")
        print(" 6. Registrar nuevo usuario y conexiones [Punto 6]")
        print(" 7. Buscar ficha detallada de un usuario específico")
        print(" 8. Salir del sistema")
        print("════════════════════════════════════════════════════════")
        opcion = input("Digite el número de la opción deseada (1-8): ").strip()
        
        if opcion == "1":
            app.mostrar_red()
        elif opcion == "2":
            app.bfs("Ana")
        elif opcion == "3":
            app.dfs("Ana")
        elif opcion == "4":
            app.menu_seleccion_usuario()
        elif opcion == "5":
            app.menu_ruta_minima()
        elif opcion == "6":
            app.menu_agregar_usuario()
        elif opcion == "7":
            app.buscar_detalles_usuario()
        elif opcion == "8":
            print("\n👋 Saliendo del sistema interactivo. ¡Mucho éxito con tu entrega del taller!\n")
            break
        else:
            print("\n[⚠️ OPCIÓN INVÁLIDA] Por favor, ingrese un número válido entre 1 y 8.\n")
 
if __name__ == "__main__":
    iniciar_consola_sistema()
```
